Project Title: RPS Arena – Rock, Paper, Scissors Game

Project Description:

RPS Arena is a Python-based Rock, Paper, Scissors game where a user competes against the computer. The computer randomly selects its choice, and the winner is determined using predefined game rules. The application supports multiple rounds, input validation, score tracking, and a final match summary.

Technologies Used:
- Python
- Random Module
- Functions
- Conditional Statements
- Loops

```python
def get_user_choice():

    while True:
        user_choice = input(
            "\nEnter rock, paper, scissors or 'quit' to exit: "
        ).lower().strip()

        if user_choice in choices or user_choice == "quit":
            return user_choice

        print("Invalid choice! Please try again.")
```

```python
player = get_user_choice()

print("You selected:", player)
```

```python
def determine_winner(user_choice, computer_choice):

    if user_choice == computer_choice:
        return "draw"

    elif (
        (user_choice == "rock" and computer_choice == "scissors") or
        (user_choice == "paper" and computer_choice == "rock") or
        (user_choice == "scissors" and computer_choice == "paper")
    ):
        return "user"

    else:
        return "computer"
```

```python
print(determine_winner("rock", "scissors"))
print(determine_winner("paper", "rock"))
print(determine_winner("scissors", "rock"))
print(determine_winner("rock", "rock"))
```

```python
def display_scoreboard(user_score, computer_score, draws):

    print("\n========== RPS ARENA SCOREBOARD ==========")

    print("Player Wins:", user_score)
    print("Computer Wins:", computer_score)
    print("Total Draws:", draws)

    print("==========================================")
```

```python
display_scoreboard(3, 2, 1)
```

```python

def welcome():

    print("\n" + "=" * 45)
    print("          WELCOME TO RPS ARENA")
    print("=" * 45)

    print("\n       ROCK  |  PAPER  |  SCISSORS")

    print("\nHow to play:")
    print("1. Choose rock, paper, or scissors.")
    print("2. The computer will randomly select its choice.")
    print("3. The winner will receive one point.")
    print("4. Enter 'quit' whenever you want to exit.")

    print("\n" + "=" * 45)
```

```python
def play_game():

    user_score = 0
    computer_score = 0
    draws = 0

    welcome()

    while True:

        user_choice = get_user_choice()

        if user_choice == "quit":
            break

        computer_choice = random.choice(choices)

        print("\nYour choice:", user_choice)
        print("Computer's choice:", computer_choice)

        result = determine_winner(user_choice, computer_choice)

        if result == "user":
            print("Congratulations! You win!")
            user_score += 1

        elif result == "computer":
            print("Computer wins!")

            computer_score += 1

        else:
            print("It's a draw!")
            draws += 1

        display_scoreboard(user_score, computer_score, draws)

    print("\n========== FINAL MATCH SUMMARY ==========")

    display_scoreboard(user_score, computer_score, draws)

    if user_score > computer_score:
        print("You are the overall winner!")

    elif computer_score > user_score:
        print("Computer is the overall winner!")

    else:
        print("The match is tied!")

    print("\nThanks for playing RPS ARENA!")
```

```python
play_game()
```
